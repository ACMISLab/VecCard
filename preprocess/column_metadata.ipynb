```python
import pandas as pd
import os
import random
from collections import Counter
import numpy as np
```

### common function

```python
def getValusCounts(df,col,minValue):
    value_counts = df[col].value_counts()
    filtered_counts = value_counts[value_counts >= minValue].sort_index()
    result_df = pd.DataFrame({'name': filtered_counts.index, 'num': filtered_counts.values.astype(str)})
    return result_df.to_dict('records') 
```

```python
def getValueCountsStatisticalData(data,type):
    numbers = [float(item['num']) for item in data]
    # 计算最大值、最小值、平均值
    max_value = max(numbers)
    min_value = min(numbers)
    avg_value = sum(numbers) / len(numbers)
    
    # 计算中位数
    numbers.sort()
    n = len(numbers)
    if n % 2 == 0:
        median = (numbers[n // 2 - 1] + numbers[n // 2]) / 2
    else:
        median = numbers[n // 2]
    
    # 计算众数
    mode_value = Counter(numbers).most_common(1)[0][0]
    return [type,max_value,min_value,avg_value,median,mode_value]
```

```python
def getNumColStatisticalData(data,type):
    numbers = data.tolist()
    # 计算最大值、最小值、平均值
    max_value = max(numbers)
    min_value = min(numbers)
    avg_value = sum(numbers) / len(numbers)
    
    # 计算中位数
    numbers.sort()
    n = len(numbers)
    if n % 2 == 0:
        median = (numbers[n // 2 - 1] + numbers[n // 2]) / 2
    else:
        median = numbers[n // 2]
    
    # 计算众数
    mode_value = Counter(numbers).most_common(1)[0][0]
    return [type,max_value,min_value,avg_value,median,mode_value]
```

```python
def getBoolColStatisticalData(data,type):
    my_list = data.tolist()
    # 初始化计数器
    true_count = 0
    false_count = 0
    
    # 遍历列表并统计True和False的数量
    for item in my_list:
        if item is True:
            true_count += 1
        elif item is False:
            false_count += 1
    
    # 计算占比
    total = true_count + false_count
    true_percentage = (true_count / total) * 100
    false_percentage = (false_count / total) * 100
    return [type,true_percentage,false_percentage]
```
