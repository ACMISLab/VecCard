```python
def split_csv(df, output_folder, row_limit=100000):
    # 确保输出文件夹存在
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    # 初始化变量
    count = 1
    split_files = []

    # 分割文件
    for i in range(0, df.shape[0], row_limit):
        # 根据row_limit切分DataFrame
        df_subset = df.iloc[i:i + row_limit]
        
        # 构造输出文件的完整路径
        file_name = os.path.join(output_folder, f"part_{count}.csv")  # 使用4位数的计数器，不足的前面补0
        # 将子集DataFrame存储为CSV文件，不包含索引
        df_subset.to_csv(file_name, index=False)
        
        # 将文件名添加到列表中
        split_files.append(file_name)
        
        # 更新计数器
        count += 1

    # 返回分割文件的列表（可选）
    return split_files
```

```python
output_folder = "/remote-home/cs_acmis_wsf/ai4dingo/yelp/vector/split_file/review"
split_files = split_csv(df, output_folder)
print("分割的CSV文件列表：", split_files)
```
