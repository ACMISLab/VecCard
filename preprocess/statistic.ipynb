```python
import json
import pandas as pd
import numpy as np
```

### Common Function

```python
def custom_function(x):
    if '_' in x:
        return int(x.split("_")[1])
    else:
        return x
```

```python
def handle_names(x):
    x = str(x)
    x = x.replace("\n","")
    x = x.replace("\t","")
    return x;
```

```python
def get_yaer(x):
    return int(str(x).split(".")[0])
```

```python
def handle_id(x):
    x = str(x)
    x = x.replace("<=","")
    return int(x);
```
