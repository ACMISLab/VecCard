```python
def split_dataframe(df, train_ratio=0.8):
    train_size = int(train_ratio * len(df))
    train_df = df[:train_size]
    test_df = df[train_size:]
    return train_df, test_df
```

```python
train_df, test_df = split_dataframe(df_5,0.8)
```

```python
train_df
```

```python
test_df
```

```python
def get_table_name(x):
    data = json.loads(x)
    return data["table"]
```

```python
test_df['table'] = test_df['cardinality'].apply(get_table_name)
```

```python
test_df
```

```python
def split_dataframe_by_column(df, column_name, target_value):
    target_df = df[df[column_name] == target_value]
    other_df = df[df[column_name] != target_value]
    return target_df, other_df
```

```python
mooccubex_test_df, yelp_test_df = split_dataframe_by_column(test_df,'table','problem')
```

```python
mooccubex_test_df  =  mooccubex_test_df.drop('table', axis=1)
```
