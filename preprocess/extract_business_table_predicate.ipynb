```python
import pandas as pd
import os
import random
from collections import Counter
import numpy as np
import json
```

```python
df = pd.read_csv("F:\\LCE\\predicate\\predicates_cardinality_all.csv")
```

```python
def get_table_name(x):
    data = json.loads(x)
    return data["table"]
```

```python
df['table'] = df['cardinality'].apply(get_table_name)
```

```python
condition = df['table'] == 'business'
```

```python
df_1 = df[condition]
```

```python
df_1 = df_1.head(5000)
```

```python
df_1
```

```python
def count_and_or(expr):
    and_count = expr.count(" and ")
    or_count = expr.count(" or ")
    return and_count + or_count
```

```python
def computer_predicate(x):
    return count_and_or(x)+1
```

```python
df_1['predicates'] = df_1['expression'].apply(computer_predicate)
```

```python
df_1
```

```python
df_1  =  df_1.drop('index', axis=1)
```

```python
df_1 = df_1.reset_index(drop=True)
```

```python
short_df = df_1[(df_1['predicates'] >= 1) & (df_1['predicates'] <= 3)]
```

```python
middle_df = df_1[(df_1['predicates'] >= 4) & (df_1['predicates'] <= 6)]
```

```python
long_df = df_1[(df_1['predicates'] >= 7)]
```

```python
short_df = short_df.reset_index(drop=True)
```

```python
middle_df = middle_df.reset_index(drop=True)
```

```python
long_df = long_df.reset_index(drop=True)
```
