```python
import pandas as pd
import numpy as np
```

```python
user_df = pd.read_csv('/media/wsf/Disk/mooccubex/csv/user.csv')
```

```python
user_df = user_df.dropna(subset=["name"])
```

```python
user_df = user_df.dropna(subset=["gender"])
```

```python
user_df
```

```python
def handle_gender(x):
    return int(x)
```

```python
user_df["gender"] = user_df["gender"].apply(handle_gender)
```

```python
def hanlde_course_order(x):
    x = str(x)
    x =  x.replace("[","")
    x= x.replace("]","")
    return x
```

```python
user_df["course_order"] = user_df["course_order"].apply( hanlde_course_order)
```

```python
def hanlde_enroll_time(x):
    x = str(x)
    x =  x.replace("[","")
    x= x.replace("]","")
    x = x.replace("'","")
    return x
```

```python
user_df
```

```python
def handle_year(x):
    if not pd.isna(x):
        x = str(x)
        x = x.split(".")[0]
        return x;
    else:
        return x;
```

```python
user_df["year_of_birth"]  = user_df["year_of_birth"].apply(handle_year)
```

### 分成多份来存储

```python
# 将数据分成多份，每份 20w 条
df = user_df
num_chunks = 12
chunk_size = 200000
chunks = [df.iloc[i*chunk_size:(i+1)*chunk_size] for i in range(num_chunks)]

# 将每份数据存成 CSV 文件
for i, chunk in enumerate(chunks):
    chunk.to_csv(f'chunk_{i+1}.csv', index=False)
```
