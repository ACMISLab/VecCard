### common

```python
def getValueCounts(df, col, val):
    value_counts = df[col].value_counts()
    filtered_counts = value_counts[value_counts >= val].sort_index()
    result_df = pd.DataFrame({'name': filtered_counts.index, 'num': filtered_counts.values.astype(str)})
    formatted_result = result_df.to_dict('records')
    print(len(formatted_result))
    print(formatted_result)    
```

### Movie

```python
movie_df = pd.read_csv('F:\\moviedata-10m\\电影数据集-CSV格式\\movies.csv')
```

```python
movie_df = movie_df.drop(['ALIAS','ACTORS','COVER','DIRECTORS','GENRES','IMDB_ID','OFFICIAL_SITE','STORYLINE','SLUG','ACTOR_IDS','DIRECTOR_IDS'],axis=1)
```

```python
movie_df = movie_df.drop(['RELEASE_DATE','TAGS'],axis=1)
```

```python
movie_df = movie_df.dropna()
```

```python
def handle_language(x):
    x = str(x)
    return x.split("/")[0]
```

```python
movie_df["LANGUAGES"] = movie_df["LANGUAGES"].apply(handle_language)
```

```python
movie_df["REGIONS"] = movie_df["REGIONS"].apply(handle_language)
```

```python
 movie_df["YEAR"] = movie_df["YEAR"].astype(int)
```

```python
 movie_df["DOUBAN_VOTES"] = movie_df["DOUBAN_VOTES"].astype(int)
```

```python
movie_df = movie_df.rename(columns={'MOVIE_ID': 'movie_id', 'NAME': 'name','DOUBAN_SCORE':"socre","LANGUAGES":"language","MINS":"mins","REGIONS":"region","YEAR":"year"})
```

```python
movie_df = movie_df.rename(columns={'DOUBAN_VOTES': 'vote'})
```

```python
movie_df.to_csv('F:\\LCE\\dataset\\douban\\movies.csv', index=False)
```
